```python
import hashlib
import json
import time
from typing import Dict
```

```python
class Block:
    def __init__(self, index, timestamp, data, signature, public_key, previous_hash):
        self.index = index
        self.timestamp = timestamp
        self.data = data              # Estado del gemelo (JSON)
        self.signature = signature    # Firma del estado
        self.public_key = public_key  # Identidad del gemelo
        self.previous_hash = previous_hash
        self.hash = self.compute_hash()

    def compute_hash(self):
        block_string = json.dumps({
            "index": self.index,
            "timestamp": self.timestamp,
            "data": self.data,
            "signature": self.signature,
            "public_key": self.public_key,
            "previous_hash": self.previous_hash
        }, sort_keys=True)

        return hashlib.sha256(block_string.encode()).hexdigest()
```

```python
class Blockchain:
    def __init__(self):
        self.chain = []
        self.create_genesis_block()

    def create_genesis_block(self):
        genesis_block = Block(
            index=0,
            timestamp=time.time(),
            data={"genesis": True},
            signature="",
            public_key="SYSTEM",
            previous_hash="0"
        )
        self.chain.append(genesis_block)

    @property
    def last_block(self):
        return self.chain[-1]

    def add_gemelo_state(self, data: Dict, signature: str, public_key: str):
        block = Block(
            index=len(self.chain),
            timestamp=time.time(),
            data=data,
            signature=signature,
            public_key=public_key,
            previous_hash=self.last_block.hash
        )
        self.chain.append(block)

    def is_chain_valid(self):
        for i in range(1, len(self.chain)):
            current = self.chain[i]
            previous = self.chain[i - 1]

            if current.hash != current.compute_hash():
                return False

            if current.previous_hash != previous.hash:
                return False

        return True
```

```python
blockchain = Blockchain()

blockchain.add_gemelo_state(
    data={"temp": 36.7, "estado": "ok"},
    signature="firma_fake",
    public_key="gemelo_001"
)

print("Cadena válida:", blockchain.is_chain_valid())
print("Bloques:", len(blockchain.chain))
```
